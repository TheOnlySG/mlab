```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
path = 'datasets/Mall_Customers.csv'
data = pd.read_csv(path)
```

```python
data.head()
```

```python
x = data[['Age' , 'Annual Income (k$)' ,'Spending Score (1-100)']]
x.head()
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled[:5]
```

```python
model = DBSCAN(
    min_samples = 3,
    eps = 0.6
)
```

```python
labels = model.fit_predict(x_scaled)
labels
```

```python
data['clusters'] = labels
```

```python
data.head()
```

```python
data['clusters'].value_counts()
```

```python
sns.scatterplot(data = data , x = 'Annual Income (k$)' , y='Spending Score (1-100)',
               # hue = 'cluster' ,
               # palette='tab10',
               s=100,
               # label = 'datapoint'
               )

clustered = data[data['clusters']!=-1]
plt.scatter(
    clustered['Annual Income (k$)'],
    clustered['Spending Score (1-100)']
    , c=clustered['clusters'],
    cmap='viridis',
    s=100,
    # label='clusters'
)
noise=data[data['clusters']==-1]
plt.scatter(
    noise['Annual Income (k$)'],
    noise['Spending Score (1-100)'],
    color='red',
    s=100,
    label='outliers'
)

plt.legend()
```
